```python
%%sql
SELECT
TIME_FORMAT(TIME(time_of_record), '%H:00') AS hour_of_day, DAYNAME(time_of_record),
CASE
WHEN DAYNAME(time_of_record) = 'Sunday' THEN time_in_queue ELSE NULL END AS Sunday
FROM visits WHERE
time_in_queue != 0
LIMIT 5;
```

```python
%%sql
SELECT
TIME_FORMAT(TIME(time_of_record), '%H:00') AS hour_of_day,
ROUND(AVG(CASE WHEN DAYNAME(time_of_record) = 'Sunday' THEN time_in_queue ELSE NULL END),0) AS Sunday,
ROUND(AVG(CASE WHEN DAYNAME(time_of_record) = 'Monday' THEN time_in_queue ELSE NULL END),0) AS Monday,
ROUND(AVG(CASE WHEN DAYNAME(time_of_record) = 'Tuesday' THEN time_in_queue ELSE NULL END),0) AS Tuesday,
ROUND(AVG(CASE WHEN DAYNAME(time_of_record) = 'Wednesday' THEN time_in_queue ELSE NULL END),0) AS Wednesday,
ROUND(AVG(CASE WHEN DAYNAME(time_of_record) = 'Thursday' THEN time_in_queue ELSE NULL END),0) AS Thursday,
ROUND(AVG(CASE WHEN DAYNAME(time_of_record) = 'Friday' THEN time_in_queue ELSE NULL END),0) AS Friday,
ROUND(AVG(CASE WHEN DAYNAME(time_of_record) = 'Saturday' THEN time_in_queue ELSE NULL END),0) AS Saturday
FROM visits
WHERE
time_in_queue != 0-- this excludes other sources with 0 queue times
GROUP BY hour_of_day
ORDER BY hour_of_day;
```

#### Joining auditor_report, visit table and water_quality 

```python
%%sql
SELECT 
  visits.record_id,  
  auditor_report.true_water_source_score AS auditor_score, 
  visits.location_id AS visit_location,  
  auditor_report.location_id AS audit_location,
  water_quality.subjective_quality_score AS survey_score
FROM 
  auditor_report
JOIN 
  visits 
ON 
  auditor_report.location_id = visits.location_id
JOIN 
  water_quality 
ON 
  visits.visit_count = water_quality.visit_count 
AND 
  visits.record_id = water_quality.record_id
WHERE 
  auditor_report.true_water_source_score = water_quality.subjective_quality_score
AND visits.visit_count = 1
LIMIT 5;
```

```python
%%sql
WITH error_count AS (
SELECT
  employee_name,
  COUNT(*) AS number_of_mistakes
FROM
  md_water_services.incorrect_record
GROUP BY 
  employee_name
ORDER BY 
  number_of_mistakes DESC)
SELECT AVG(number_of_mistakes) AS average_mistake FROM error_count;
```

```python
%%sql
SELECT 
    employee_name,
    COUNT(*) AS number_of_mistakes
FROM 
    md_water_services.incorrect_record
GROUP BY 
    employee_name
HAVING 
    COUNT(*) > (
        SELECT AVG(mistake_count)
        FROM (
            SELECT COUNT(*) AS mistake_count
            FROM md_water_services.incorrect_record
            GROUP BY employee_name
        ) AS subquery
    )
ORDER BY 
    number_of_mistakes DESC;
```

```python
%%sql
CREATE OR REPLACE VIEW md_water_services.suspect_list AS
SELECT 
    employee_name,
    COUNT(*) AS number_of_mistakes
FROM 
    md_water_services.incorrect_record
GROUP BY 
    employee_name
HAVING 
    COUNT(*) > (
        SELECT AVG(mistake_count)
        FROM (
            SELECT COUNT(*) AS mistake_count
            FROM md_water_services.incorrect_record
            GROUP BY employee_name
        ) AS subquery
    )
ORDER BY 
    number_of_mistakes DESC;
```
